```python
#Importing python libraries 
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#Dirty Values - NA,Null,Blank,spaces
df.replace(
    ['N/A','Null','',' '],np.nan,inplace=True
)
```

```python
df.isnull().sum()
```

```python
#Standardize Text 
df["type"] = df["type"].str.title()
df["title"] = df["title"].str.title()
df["director"] = df["director"].str.title()
df["cast"] = df["cast"].str.title()
df["country"] = df["country"].str.title()
df["listed_in"] = df["listed_in"].str.title()
df["description"] = df["description"].str.title()
```

```python
df["country"].value_counts()
```

```python
df.describe()
```

```python
# Cleaning data

df["date_added"] = pd.to_datetime(
    df["date_added"],
    format="mixed",
    errors="coerce"
)

# Handling Missing Values
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["duration"] = df["duration"].fillna("Unknown")
```

```python
#Handling Missing value of rating 
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
```

```python
#Feature Engineering
#Added Year
df["added_year"] = df["date_added"].dt.year
#Added Month
df["added_month"] = df["date_added"].dt.month_name()
#Content Age
df["content_age"] = df["added_year"] - df["release_year"]
#How recent is the content ?
df["is_recent"] = df["content_age"].apply(
    lambda x: "Recent" if 0 <= x <= 2
    else "Older" if x > 2
    else "Unknown"
)
#Duration Value 
df["duration_value"] = (
    df["duration"]
    .str.extract(r"(\d+)", expand=False)
    .astype(float)
)
#Duration Type
df["duration_type"] = (
    df["duration"]
    .str.extract(r"([A-Za-z]+)", expand=False)
)

df["duration_type"] = df["duration_type"].replace({
    "min": "Minutes",
    "Season": "Seasons",
    "Seasons": "Seasons"
})
#Genre Count
df["genre_count"] = df["listed_in"].str.split(",").str.len()
#Country Count
df["country_count"] = df["country"].str.split(",").str.len()
```

```python
#Univariate Analysis 
#Question : How many movies and TV shows are there ?
df["type"].value_counts()
```

```python
#Visualization
df["type"].value_counts().plot(kind = 'bar')
font1 = {'family':'serif', 'color' : 'darkblue' , 'size' : 20}
font2 = {'family':'serif', 'color' : 'darkred' , 'size' : 12}
plt.title("Number of Movies and TV shows", fontdict = font1)
plt.xlabel("Type", fontdict = font2)
plt.ylabel("Number of titles", fontdict = font2)
plt.show()

#Insights : Movies make up the majority of Netflix's content catalog, while TV Shows represent a smaller share. This indicates that Netflix's library has historically been more heavily focused on movie content.
```

```python
#Analyzing rating 
#Question : What are the most common ratings on Netflix?
df['rating'].value_counts()
```

```python
# Visualization
df["rating"].value_counts().plot(kind = 'bar')
font1 = {'family':'serif', 'color' : 'darkblue' , 'size' : 20}
font2 = {'family':'serif', 'color' : 'darkred' , 'size' : 12}
plt.title("Netflix Rating", fontdict = font1)
plt.xlabel("Rating", fontdict = font2)
plt.ylabel("Number of titles", fontdict = font2)
plt.xticks(rotation=45)
plt.show()

#Insight : Netflix's catalog is concentrated around a few common content ratings. This suggests that a large portion of the platform's content is targeted toward general, teen, and mature audiences rather than being evenly distributed across all rating categories.
```

```python
#Analyzing
#Question : How do ratings differ between Movies and TV Shows?
rating_type = pd.crosstab(df['rating'],df['type'])
rating_type
```

```python
#Visualizing movies vs TV shows rating 
rating_type.plot(kind = 'bar' , figsize = (12,6))
font1 = {'family':'serif', 'color':'darkblue', 'size':20}
font2 = {'family':'serif', 'color':'darkred', 'size':12}

plt.title("Netflix Content by Rating", fontdict=font1)
plt.xlabel("Rating", fontdict=font2)
plt.ylabel("Number of Titles", fontdict=font2)
plt.xticks(rotation=45)
plt.legend(title="Type")
plt.show()

#Inshight : The distribution of ratings varies between Movies and TV Shows. Some ratings are more prevalent among Movies, while others have a stronger presence among TV Shows, indicating differences in the type of audiences and content represented in each category.
```

```python
#Analyzing
#How has the amount of content added to the netflix changed over the time?
df['added_year'].value_counts().sort_index()
```

```python
#Visulaization 
content_by_year = df['added_year'].value_counts().sort_index()
plt.figure(figsize = (12,6))
plt.plot(content_by_year.index , content_by_year.values, marker = 'o')
font1 = {'family':'serif', 'color':'darkblue', 'size':20}
font2 = {'family':'serif', 'color':'darkred', 'size':12}

plt.title("Netflix Content Added Over the Years", fontdict = font1)
plt.xlabel("Years", fontdict = font2)
plt.ylabel("Number of Titles", fontdict = font2)
plt.grid(True, alpha = 0.5)
plt.show()

#Inshight : Netflix's content library has expanded considerably over time, with the dataset showing a strong increase in titles during the later years. This reflects Netflix's rapid expansion as a global streaming platform and its increasing investment in original and acquired content.
```
